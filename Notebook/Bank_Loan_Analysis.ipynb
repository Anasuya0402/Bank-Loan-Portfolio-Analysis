```python

import pandas as pd
from pathlib import Path

project_folder = Path.cwd()

if project_folder.name.lower() == "notebook":
    project_folder = project_folder.parent

file_path = project_folder / "Data" / "cleaned_financial_loan.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python

from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

# Connect to the Bank Loan Analysis database
database_url = URL.create(
    "postgresql+psycopg2",
    username=username,
    password=password,
    host="127.0.0.1",
    port=5432,
    database="bank_loan_analysis"
)

engine = create_engine(
    database_url,
    connect_args={"connect_timeout": 5}
)

# Upload the cleaned dataset
df.to_sql(
    "cleaned_financial_loan",
    engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)

# Verify the uploaded rows
with engine.connect() as conn:
    uploaded_rows = conn.execute(
        text("SELECT COUNT(*) FROM cleaned_financial_loan")
    ).scalar()

print("Dataset uploaded successfully!")
print("DataFrame rows:", len(df))
print("PostgreSQL rows:", uploaded_rows)
print("Row counts match:", len(df) == uploaded_rows)
```
